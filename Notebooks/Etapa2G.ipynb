```python
from abc import ABC, abstractmethod
# ==========================================
# INTERFACES E CLASSES ABSTRATAS (SOLID: ISP)
# ==========================================
class IExibivel(ABC):
    """Interface que garante que a classe saiba se exibir no terminal."""
    @abstractmethod
    def exibir(self):
        pass
# ==========================================
# CLASSES DE DOMÍNIO (SOLID: SRP)
# ==========================================
class Plataforma:
    def __init__(self, nome, fabricante):
        self.__nome = nome
        self.__fabricante = fabricante
    # Encapsulamento com @property
    @property
    def nome(self): return self.__nome
    @nome.setter
    def nome(self, valor): self.__nome = valor
    @property
    def fabricante(self): return self.__fabricante
    def __del__(self):
        # Destrutor mantido conforme o seu código original
        pass
class Avaliacao(IExibivel):
    def __init__(self, nota, comentario):
        self.__nota = nota          
        self.__comentario = comentario
    @property
    def nota(self): return self.__nota
    @property
    def comentario(self): return self.__comentario
    def exibir(self):
        print(f"⭐ Nota: {self.nota}/10 — {self.comentario}")
class Jogo(IExibivel):
    """Classe Abstrata. Não deve ser instanciada diretamente."""
    total_jogos = 0
    def __init__(self, titulo, genero, plataforma, horas_jogadas=0):
        self.__titulo = titulo
        self.__genero = genero
        self.__plataforma = plataforma   
        self.__horas_jogadas = horas_jogadas
        self.__zerado = False
        self.__avaliacao = None          
        Jogo.total_jogos += 1
    @property
    def titulo(self): return self.__titulo
    @property
    def genero(self): return self.__genero
    @property
    def plataforma(self): return self.__plataforma
    @property
    def horas_jogadas(self): return self.__horas_jogadas
    @property
    def zerado(self): return self.__zerado
    @zerado.setter
    def zerado(self, valor): self.__zerado = valor
    @property
    def avaliacao(self): return self.__avaliacao
    @avaliacao.setter
    def avaliacao(self, av): self.__avaliacao = av
    # Método abstrato: as classes filhas DEVEM implementar (Herança e Polimorfismo)
    @abstractmethod
    def exibir(self):
        pass
# ==========================================
# HERANÇA E POLIMORFISMO (SOLID: OCP e LSP)
# ==========================================
class JogoDigital(Jogo):
    def __init__(self, titulo, genero, plataforma, tamanho_gb, horas_jogadas=0):
        super().__init__(titulo, genero, plataforma, horas_jogadas)
        self.__tamanho_gb = tamanho_gb
    def exibir(self):
        status = "✅ Zerado" if self.zerado else "⚠️ Em andamento"
        print(f"☁️ [Digital] {self.titulo} | {self.genero} | {self.plataforma.nome} | {self.__tamanho_gb}GB | {self.horas_jogadas}h | {status}")
        if self.avaliacao:
            self.avaliacao.exibir()
class JogoFisico(Jogo):
    def __init__(self, titulo, genero, plataforma, possui_caixa=True, horas_jogadas=0):
        super().__init__(titulo, genero, plataforma, horas_jogadas)
        self.__possui_caixa = possui_caixa
    def exibir(self):
        status = "✅ Zerado" if self.zerado else "⚠️ Em andamento"
        caixa = "📦 Na caixa" if self.__possui_caixa else "💿 Só a mídia"
        print(f"💿 [Físico] {self.titulo} | {self.genero} | {self.plataforma.nome} | {caixa} | {self.horas_jogadas}h | {status}")
        if self.avaliacao:
            self.avaliacao.exibir()
# ==========================================
# GERENCIAMENTO E ASSOCIAÇÃO
# ==========================================
class Biblioteca:
    def __init__(self):
        self.__jogos = []
    @property
    def jogos(self): return self.__jogos
    def adicionar_jogo(self, jogo: Jogo):
        self.__jogos.append(jogo)
```
