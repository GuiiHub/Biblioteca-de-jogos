# Sistema de Biblioteca de Jogos 🎮
________________________________________________________________________________________________________________________________________________________

## Contexto / Persona
Lucas, 22 anos, gamer casual que tem mais de 80 jogos espalhados em Steam, PS5 e Xbox. Ele nunca sabe o que jogar, esquece quais jogos já zerou e não consegue recomendar nada para os amigos. Quer um sistema para catalogar, avaliar e organizar sua coleção.

## Classes


```python
'''
Jogo                → dados de cada jogo
Plataforma          → Steam, PS5, Xbox...
Avaliacao           → nota e comentário do usuário
Biblioteca          → coleção completa do usuário
Usuario             → dono da biblioteca
'''
```

# Código Principal

```python
class Plataforma:
    def __init__(self, nome, fabricante):
        self.__nome = nome
        self.__fabricante = fabricante

    def get_nome(self): return self.__nome
    def set_nome(self, v): self.__nome = v
    def get_fabricante(self): return self.__fabricante
    def set_fabricante(self, v): self.__fabricante = v

    def __del__(self):
        print(f"Plataforma {self.__nome} removida.")


class Avaliacao:
    def __init__(self, nota, comentario):
        self.__nota = nota          
        self.__comentario = comentario

    def get_nota(self): return self.__nota
    def set_nota(self, v): self.__nota = v
    def get_comentario(self): return self.__comentario
    def set_comentario(self, v): self.__comentario = v

    def exibir(self):
        print(f"⭐ Nota: {self.__nota}/10 — {self.__comentario}")


class Jogo:
    total_jogos = 0

    def __init__(self, titulo, genero, plataforma, horas_jogadas=0):
        self.__titulo = titulo
        self.__genero = genero
        self.__plataforma = plataforma   
        self.__horas_jogadas = horas_jogadas
        self.__zerado = False
        self.__avaliacao = None          
        Jogo.total_jogos += 1

    def get_titulo(self): return self.__titulo
    def set_titulo(self, v): self.__titulo = v
    def get_genero(self): return self.__genero
    def set_genero(self, v): self.__genero = v
    def get_horas(self): return self.__horas_jogadas
    def set_horas(self, v): self.__horas_jogadas = v
    def get_zerado(self): return self.__zerado
    def set_zerado(self, v): self.__zerado = v
    def get_avaliacao(self): return self.__avaliacao
    def set_avaliacao(self, av): self.__avaliacao = av
    def get_plataforma(self): return self.__plataforma

    def exibir(self):
        status = "✅ Zerado" if self.__zerado else "⚠️ Em andamento"
        print(f"{self.__titulo} | {self.__genero} | "
              f"{self.__plataforma.get_nome()} | "
              f"{self.__horas_jogadas}h | {status}")
        if self.__avaliacao:
            self.__avaliacao.exibir()

    def __del__(self):
        print(f"Jogo '{self.__titulo}' removido da biblioteca.")


class Biblioteca:
    def __init__(self):
        self.__jogos = []

    def adicionar_jogo(self, jogo):
        self.__jogos.append(jogo)
        print(f"'{jogo.get_titulo()}' adicionado!")

    def remover_jogo(self, titulo):
        for j in self.__jogos:
            if j.get_titulo() == titulo:
                self.__jogos.remove(j)
                return
        print("Jogo não encontrado.")

    def listar_todos(self):
        if not self.__jogos:
            print("Biblioteca vazia.")
            return
        for j in self.__jogos:
            j.exibir()


    def total_horas(self):
        return sum(j.get_horas() for j in self.__jogos)

    def listar_zerados(self):
        zerados = [j for j in self.__jogos if j.get_zerado()]
        for j in zerados:
            j.exibir()

    def listar_em_andamento(self):
        em_andamento = [j for j in self.__jogos if not j.get_zerado()]
        
        if not em_andamento:
            print("Nenhum jogo em andamento no momento.")
            return
            
        for j in em_andamento:
            j.exibir()


    def buscar_por_genero(self, genero):
        resultado = [j for j in self.__jogos if j.get_genero() == genero]
        for j in resultado:
            j.exibir()

    def media_horas(self):
        if not self.__jogos:
            return 0
        total = sum(j.get_horas() for j in self.__jogos)
        return total / len(self.__jogos)

    def get_jogos(self): return self.__jogos


class Usuario:
    total_usuarios = 0  # atributo de classe

    def __init__(self, nome, username):
        self.__nome = nome
        self.__username = username
        self.__biblioteca = Biblioteca()
        Usuario.total_usuarios += 1

    def get_nome(self): return self.__nome
    def set_nome(self, v): self.__nome = v
    def get_username(self): return self.__username
    def get_biblioteca(self): return self.__biblioteca

    def exibir_perfil(self):
        total_horas = self.__biblioteca.total_horas()
    
        total_na_biblioteca = len(self.__biblioteca.get_jogos()) 
        
        media = self.__biblioteca.media_horas()
        print(f"👤 {self.__nome} (@{self.__username})")
        print(f"🎮 Total de jogos na biblioteca: {total_na_biblioteca}")
        print(f"⏱️ Total de horas: {total_horas}h") 
        print(f"⏱️ Média de horas jogadas: {media:.1f}h")

    def __del__(self):
        print(f"Usuário {self.__username} removido.")

#─────────────────────────────────────────────────────────────────────────
#─── EXECUÇÃO PRINCIPAL ─── Testar as classes ────────────────────────────
if __name__ == "__main__":
    ps5 = Plataforma("PS5", "Sony")
    steam = Plataforma("Steam", "Valve")
    xbox = Plataforma("Xbox", 'Microsoft')
    nintendo = Plataforma("Nintendo Switch", "Nintendo")

# ---------------------- Regsistar Usuário -------------------------------

    lucas = Usuario("Lucas Mendes", "lucasgamer")

# ---------------------- Regsistar os jogos -------------------------------
    j1 = Jogo("Hollow Knight", "Metroidvania", steam, 40)
    j1.set_zerado(True)
    j1.set_avaliacao(Avaliacao(9, "Obra de arte, vale cada centavo."))

    j2 = Jogo("God of War", "Ação/Aventura", ps5, 25)
    j2.set_zerado(False)

    j3 = Jogo("Hades", "Roguelike", steam, 60)
    j3.set_zerado(True)
    j3.set_avaliacao(Avaliacao(10, "Impossível parar de jogar."))

    j4 = Jogo("Resident Evil 4", "Survival Horror", xbox, 40)
    j4.set_zerado(False)


# ------ Adicionar os jogos registrados -------
# Usar o método adicionar_jogo()
    lib = lucas.get_biblioteca()
    lib.adicionar_jogo(j1)
    lib.adicionar_jogo(j2)
    lib.adicionar_jogo(j3)
    lib.adicionar_jogo(j4)

# ------- Remover jogos ----------
#  Usar o método remover_jogo() e chamar o destrutor del


# ──────────────────────────────────

    print("\n=== Todos os jogos ===")
    lib.listar_todos()

    print("\n==⚠️==⚠️==⚠️== Jogos em andamento ==⚠️==⚠️==⚠️==")
    lib.listar_em_andamento()

    print("\n==✅==✅==✅== Jogos zerados ==✅==✅==✅==")
    lib.listar_zerados()

    print("\n=== Perfil do usuário ===")
    lucas.exibir_perfil()
```

# Diagrama de classes

```python
'''
Usuario
├── total_usuarios (classe)
├── __nome, __username
├── __biblioteca → Biblioteca
└── exibir_perfil()

Biblioteca
├── __jogos → [Jogo]
├── adicionar/remover_jogo()
├── listar_todos/zerados()
├── buscar_por_genero()
└── media_horas()

Jogo
├── total_jogos (classe)
├── __titulo, __genero, __horas_jogadas, __zerado
├── __plataforma → Plataforma
├── __avaliacao  → Avaliacao
└── exibir()

Plataforma              Avaliacao
├── __nome              ├── __nota
├── __fabricante        ├── __comentario
└── get/set...          └── exibir()
'''
```
